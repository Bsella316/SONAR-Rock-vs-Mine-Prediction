#**SONAR Rock vs Mine Prediction with Python**

##**1. Problem Definition:**
This project aims to develop a supervised Machine Learning classification system capable of distinguishing between underwater rocks and mines using SONAR signal data. The dataset consists of 208 sonar observations, with each observation represented by 60 numerical features describing the energy of the returned sonar signal across different frequency bands. The model will learn the patterns associated with rocks and metal mines and classify new sonar signals into two classes: Rock (R) or Mine (M). The project will involve data exploration, preprocessing, model training, evaluation, and comparison of classification algorithms to determine an effective model for the task.

##**2. Data Understanding:**
The 60 columns are not different physical properties of the rocks or mines. Instead, they represent measurements extracted from the returned SONAR signal.

Each value represents the energy of the reflected sonar signal within a particular frequency band, with values ranging from 0.0 to 1.0

##**7. Model Evaluation and Comparison**

####**a) Logistic Regression Accuracy**

```python
# Accuracy on training data
X_train_prediction = model.predict(X_train)
training_data_accuracy = accuracy_score(X_train_prediction, Y_train)
print('Accuracy on training data for Logistic Regression:', training_data_accuracy)

X_test_prediction = model.predict(X_test)
test_data_accuracy = accuracy_score(X_test_prediction, Y_test)
print('Accuracy on test data for Logistic Regression:', test_data_accuracy)
```

####**b) SVM Accuracy**

```python
# Accuracy on training data
X_train_prediction_svm = svm_model.predict(X_train)
training_data_accuracy_svm = accuracy_score(X_train_prediction_svm, Y_train)
print('Accuracy on training data for SVM:', training_data_accuracy_svm)

X_test_prediction_svm = svm_model.predict(X_test)
test_data_accuracy_svm = accuracy_score(X_test_prediction_svm, Y_test)
print('Accuracy on test data for SVM:', test_data_accuracy_svm)
```

```python
knn_model = KNeighborsClassifier()
knn_model.fit(X_train, Y_train)

# Accuracy on training data
X_train_prediction_knn = knn_model.predict(X_train)
training_data_accuracy_knn = accuracy_score(X_train_prediction_knn, Y_train)
print('Accuracy on training data for KNN:', training_data_accuracy_knn)

X_test_prediction_knn = knn_model.predict(X_test)
test_data_accuracy_knn = accuracy_score(X_test_prediction_knn, Y_test)
print('Accuracy on test data for KNN:', test_data_accuracy_knn)
```

#### **d) Decision Tree Accuracy**

```python
# Accuracy on training data
X_train_prediction_dtree = dtree_model.predict(X_train)
training_data_accuracy_dtree = accuracy_score(X_train_prediction_dtree, Y_train)
print('Accuracy on training data for Decision Tree:', training_data_accuracy_dtree)

X_test_prediction_dtree = dtree_model.predict(X_test)
test_data_accuracy_dtree = accuracy_score(X_test_prediction_dtree, Y_test)
print('Accuracy on test data for Decision Tree:', test_data_accuracy_dtree)
```

####**e) Random Forest Accuracy**

```python
# Accuracy on training data
X_train_prediction_rf = rf_model.predict(X_train)
training_data_accuracy_rf = accuracy_score(X_train_prediction_rf, Y_train)
print('Accuracy on training data for Random Forest:', training_data_accuracy_rf)

X_test_prediction_rf = rf_model.predict(X_test)
test_data_accuracy_rf = accuracy_score(X_test_prediction_rf, Y_test)
print('Accuracy on test data for Random Forest:', test_data_accuracy_rf)
```

####**f) Gradient Boosting Accuracy**

```python
# Accuracy on training data
X_train_prediction_gbc = gbc_model.predict(X_train)
training_data_accuracy_gbc = accuracy_score(X_train_prediction_gbc, Y_train)
print('Accuracy on training data for Gradient Boosting:', training_data_accuracy_gbc)

X_test_prediction_gbc = gbc_model.predict(X_test)
test_data_accuracy_gbc = accuracy_score(X_test_prediction_gbc, Y_test)
print('Accuracy on test data for Gradient Boosting:', test_data_accuracy_gbc)
```

##**8. Model Comparison Summary**

```python
model_comparison = pd.DataFrame({
    'Model': ['Logistic Regression', 'SVM', 'KNN', 'Decision Tree', 'Random Forest', 'Gradient Boosting'],
    'Training Accuracy': [training_data_accuracy, training_data_accuracy_svm, training_data_accuracy_knn, training_data_accuracy_dtree, training_data_accuracy_rf, training_data_accuracy_gbc],
    'Test Accuracy': [test_data_accuracy, test_data_accuracy_svm, test_data_accuracy_knn, test_data_accuracy_dtree, test_data_accuracy_rf, test_data_accuracy_gbc]
})
display(model_comparison.sort_values(by='Test Accuracy', ascending=False))
```

### **Summary of Model Performance:**

Based on the test accuracy, the **K-Nearest Neighbors (KNN)** and **Support Vector Machine (SVM)** models performed the best on this dataset, both achieving an accuracy of approximately 80.95%.

*   **Logistic Regression** and **Gradient Boosting** models also showed decent performance, with test accuracies of around 76.19%.
*   The **Random Forest** model had a test accuracy of about 71.43%.
*   The **Decision Tree** model showed signs of overfitting, with a perfect training accuracy (100%) but the lowest test accuracy (61.90%), indicating it learned the training data too well but generalized poorly to unseen data.

For practical application, **KNN** or **SVM** would be good starting points, potentially followed by hyperparameter tuning to further optimize their performance.

```python
# ============================================================
# SONAR DEEP SEA AI
# V2 - ORGANIZED + FIXED
# GOOGLE COLAB READY
# ============================================================

import gradio as gr
import numpy as np
import time

# ============================================================
# 1. CHECK MODEL
# ============================================================

if "svm_model" not in globals():
    raise Exception("svm_model was not found. Please train or load your SVM model before running this cell.")

# ============================================================
# 2. PREDICTION FUNCTION
# ============================================================

def predict_sonar(*features):
    try:
        values = []
        for value in features:
            if value is None or value == "":
                values.append(0.0)
            else:
                values.append(float(value))

        if len(values) != 60:
            return """
            <div class="error-card">
                <div class="error-icon">!</div>
                <div class="error-title">INVALID SIGNAL</div>
                <div class="error-text">The model requires exactly 60 SONAR features.</div>
            </div>
            """

        input_data = np.array(values, dtype=float).reshape(1, 60)
        prediction = svm_model.predict(input_data)[0]
        prediction = str(prediction).strip().upper()

        if prediction in ["R", "ROCK", "0"]:
            object_type = "ROCK"
            icon = "◈"
            status = "SAFE SIGNAL"
            result_class = "rock-result"
            description = "The SONAR acoustic pattern is consistent with a natural rocky object."
        else:
            object_type = "MINE"
            icon = "⚠"
            status = "WARNING SIGNAL"
            result_class = "mine-result"
            description = "The SONAR acoustic pattern is consistent with a possible underwater mine."

        confidence = None

        if hasattr(svm_model, "predict_proba"):
            try:
                probabilities = svm_model.predict_proba(input_data)[0]
                classes = list(svm_model.classes_)
                index = classes.index(svm_model.predict(input_data)[0])
                confidence = probabilities[index] * 100
            except Exception:
                confidence = None

        if confidence is None and hasattr(svm_model, "decision_function"):
            try:
                score = svm_model.decision_function(input_data)
                score = abs(float(score[0]))
                confidence = 50 + 50 * (1 / (1 + np.exp(-score)))
                confidence = min(max(confidence, 50), 99.9)
            except Exception:
                confidence = None

        if confidence is not None:
            confidence_text = f"{confidence:.2f}%"
            confidence_width = confidence
        else:
            confidence_text = "N/A"
            confidence_width = 0

        signal_mean = np.mean(values)
        signal_std = np.std(values)
        signal_min = np.min(values)
        signal_max = np.max(values)

        return f"""
        <div class="result-card {result_class}">
            <div class="result-icon">{icon}</div>
            <div class="result-small-title">SONAR CLASSIFICATION</div>
            <div class="result-title">{object_type} DETECTED</div>
            <div class="result-status">● {status}</div>
            <div class="confidence-container">
                <div class="confidence-label">MODEL CONFIDENCE</div>
                <div class="confidence-value">{confidence_text}</div>
                <div class="confidence-bar">
                    <div class="confidence-fill" style="width:{confidence_width}%"></div>
                </div>
            </div>
            <div class="statistics">
                <div class="stat-box">
                    <div class="stat-value">60</div>
                    <div class="stat-label">FEATURES</div>
                </div>
                <div class="stat-box">
                    <div class="stat-value">SVM</div>
                    <div class="stat-label">MODEL</div>
                </div>
                <div class="stat-box">
                    <div class="stat-value">{signal_mean:.3f}</div>
                    <div class="stat-label">MEAN</div>
                </div>
                <div class="stat-box">
                    <div class="stat-value">{signal_std:.3f}</div>
                    <div class="stat-label">STD</div>
                </div>
            </div>
            <div class="description">{description}</div>
            <div class="signal-range">SIGNAL RANGE: {signal_min:.3f} → {signal_max:.3f}</div>
        </div>
        """

    except Exception as error:
        return f"""
        <div class="error-card">
            <div class="error-icon">!</div>
            <div class="error-title">PREDICTION ERROR</div>
            <div class="error-text">{str(error)}</div>
        </div>
        """

# ============================================================
# 3. RANDOM SONAR SIGNAL
# ============================================================

def generate_random_signal():
    return [round(np.random.uniform(0, 1), 4) for _ in range(60)]

# ============================================================
# 4. CLEAR SIGNAL
# ============================================================

def clear_signal():
    return [0.0] * 60

# ============================================================
# 5. CUSTOM CSS
# ============================================================

css = """
body {
    background: radial-gradient(circle at 50% 0%, #07516b 0%, #023348 30%, #011c2b 60%, #000b14 100%) !important;
    color: #e5fbff !important;
}
.gradio-container {
    max-width: 1450px !important;
    margin: auto !important;
    background: transparent !important;
}
.bubbles {
    position: fixed;
    width: 100%;
    height: 100%;
    left: 0;
    top: 0;
    pointer-events: none;
    overflow: hidden;
    z-index: 0;
}
.bubble {
    position: absolute;
    bottom: -50px;
    border-radius: 50%;
    border: 1px solid rgba(100,230,255,0.3);
    animation: rise 12s infinite linear;
}
.bubble:nth-child(1) {
    width: 10px;
    height: 10px;
    left: 10%;
    animation-duration: 12s;
}
.bubble:nth-child(2) {
    width: 6px;
    height: 6px;
    left: 25%;
    animation-duration: 9s;
}
.bubble:nth-child(3) {
    width: 16px;
    height: 16px;
    left: 45%;
    animation-duration: 15s;
}
.bubble:nth-child(4) {
    width: 8px;
    height: 8px;
    left: 65%;
    animation-duration: 11s;
}
.bubble:nth-child(5) {
    width: 14px;
    height: 14px;
    left: 85%;
    animation-duration: 14s;
}
@keyframes rise {
    0% {
        transform: translateY(0);
        opacity: 0;
    }
    15% {
        opacity: 0.6;
    }
    100% {
        transform: translateY(-110vh);
        opacity: 0;
    }
}
.header {
    position: relative;
    padding: 35px;
    margin: 20px 0;
    text-align: center;
    border-radius: 25px;
    background: linear-gradient(145deg, rgba(3,65,87,0.85), rgba(0,18,30,0.95));
    border: 1px solid rgba(70,220,250,0.25);
    box-shadow: 0 20px 60px rgba(0,0,0,0.4);
}
.title {
    font-size: 48px;
    font-weight: 900;
    letter-spacing: 8px;
    background: linear-gradient(90deg, white, #61eaff, #1bbce8, white);
    -webkit-background-clip: text;
    -webkit-text-fill-color: transparent;
}
.subtitle {
    margin-top: 10px;
    color: #76c8d8;
    font-size: 13px;
    letter-spacing: 4px;
}
.radar {
    width: 210px;
    height: 210px;
    margin: 30px auto;
    border-radius: 50%;
    position: relative;
    background: radial-gradient(circle, rgba(0,255,190,0.15), transparent 70%);
    border: 2px solid rgba(70,230,255,0.45);
    box-shadow: 0 0 40px rgba(0,220,255,0.2);
}
.radar::before {
    content: "";
    position: absolute;
    inset: 25%;
    border-radius: 50%;
    border: 1px solid rgba(70,230,255,0.3);
}
.radar::after {
    content: "";
    position: absolute;
    left: 50%;
    top: 50%;
    width: 1px;
    height: 100%;
    background: rgba(70,230,255,0.25);
}
.radar-line {
    position: absolute;
    top: 50%;
    left: 50%;
    width: 50%;
    height: 2px;
    transform-origin: left center;
    background: linear-gradient(90deg, transparent, #48ffe0);
    box-shadow: 0 0 15px #48ffe0;
    animation: rotateRadar 3s linear infinite;
}
.radar-target {
    position: absolute;
    width: 8px;
    height: 8px;
    border-radius: 50%;
    background: #5affdf;
    top: 30%;
    left: 65%;
    box-shadow: 0 0 15px #5affdf;
    animation: pulse 1.5s infinite;
}
@keyframes rotateRadar {
    from {
        transform: rotate(0deg);
    }
    to {
        transform: rotate(360deg);
    }
}
@keyframes pulse {
    0%,100% {
        transform: scale(0.6);
        opacity: 0.5;
    }
    50% {
        transform: scale(1.4);
        opacity: 1;
    }
}
.status {
    display: inline-block;
    padding: 10px 20px;
    border-radius: 30px;
    background: rgba(0,255,190,0.06);
    border: 1px solid rgba(0,255,190,0.3);
    color: #55f7cf;
    font-size: 11px;
    letter-spacing: 2px;
}
.dashboard {
    display: grid;
    grid-template-columns: repeat(4,1fr);
    gap: 15px;
    margin: 20px 0;
}
.dashboard-card {
    padding: 20px;
    text-align: center;
    border-radius: 16px;
    background: rgba(3,44,61,0.75);
    border: 1px solid rgba(70,210,240,0.15);
}
.dashboard-value {
    font-size: 25px;
    font-weight: 800;
    color: #64e7fa;
}
.dashboard-label {
    margin-top: 5px;
    font-size: 9px;
    letter-spacing: 2px;
    color: #669aa6;
}
.input-panel {
    padding: 25px;
    border-radius: 22px;
    background: rgba(2,32,46,0.8);
    border: 1px solid rgba(70,210,240,0.18);
}
.signal-group {
    margin-bottom: 12px !important;
    border-radius: 14px !important;
    background: rgba(0,25,39,0.8) !important;
    border: 1px solid rgba(70,210,240,0.15) !important;
}
.feature {
    background: rgba(0,18,30,0.8) !important;
    border: 1px solid rgba(70,205,235,0.15) !important;
    border-radius: 10px !important;
}
.feature input {
    color: #eaffff !important;
}
.scan-button {
    height: 60px !important;
    margin-top: 20px;
    border-radius: 14px !important;
    background: linear-gradient(135deg, #006b88, #00a7c9) !important;
    color: white !important;
    font-weight: 800 !important;
    letter-spacing: 3px !important;
    border: 1px solid rgba(80,235,255,0.6) !important;
    box-shadow: 0 0 30px rgba(0,210,255,0.2);
}
.scan-button:hover {
    box-shadow: 0 0 45px rgba(0,220,255,0.45);
    transform: translateY(-2px);
}
.secondary-button {
    height: 48px !important;
    border-radius: 12px !important;
    background: rgba(3,50,67,0.8) !important;
    color: #8dddeb !important;
    border: 1px solid rgba(70,210,240,0.2) !important;
}
.result-card {
    margin-top: 20px;
    padding: 40px 25px;
    text-align: center;
    border-radius: 24px;
    animation: appear 0.6s ease;
}
.rock-result {
    background: radial-gradient(circle, rgba(0,255,190,0.18), rgba(0,35,42,0.95));
    border: 1px solid rgba(60,255,210,0.5);
    box-shadow: 0 0 50px rgba(0,255,190,0.12);
}
.mine-result {
    background: radial-gradient(circle, rgba(255,60,60,0.18), rgba(45,10,17,0.95));
    border: 1px solid rgba(255,80,80,0.55);
    box-shadow: 0 0 50px rgba(255,50,50,0.15);
}
.result-icon {
    font-size: 65px;
    text-shadow: 0 0 30px currentColor;
}
.result-small-title {
    margin-top: 10px;
    font-size: 10px;
    letter-spacing: 4px;
    color: #76aeb9;
}
.result-title {
    margin-top: 10px;
    font-size: 38px;
    font-weight: 900;
    letter-spacing: 6px;
}
.result-status {
    margin-top: 12px;
    font-size: 12px;
    letter-spacing: 3px;
}
.confidence-container {
    max-width: 500px;
    margin: 30px auto;
}
.confidence-label {
    font-size: 10px;
    letter-spacing: 3px;
    color: #6f9da7;
}
.confidence-value {
    margin-top: 5px;
    font-size: 32px;
    font-weight: 800;
    color: #7defff;
}
.confidence-bar {
    height: 7px;
    margin-top: 12px;
    border-radius: 20px;
    background: rgba(255,255,255,0.06);
    overflow: hidden;
}
.confidence-fill {
    height: 100%;
    border-radius: 20px;
    background: linear-gradient(90deg, #00a7c7, #55ffe0);
    box-shadow: 0 0 15px rgba(50,255,220,0.5);
}
.statistics {
    display: grid;
    grid-template-columns: repeat(4,1fr);
    gap: 10px;
    max-width: 750px;
    margin: 25px auto;
}
.stat-box {
    padding: 15px;
    border-radius: 12px;
    background: rgba(255,255,255,0.025);
    border: 1px solid rgba(255,255,255,0.06);
}
.stat-value {
    font-size: 19px;
    font-weight: 700;
    color: #8ceaf7;
}
.stat-label {
    margin-top: 5px;
    font-size: 9px;
    letter-spacing: 1.5px;
    color: #668d97;
}
.description {
    max-width: 650px;
    margin: 20px auto;
    color: #8eb6bf;
    font-size: 13px;
    line-height: 1.7;
}
.signal-range {
    color: #507d88;
    font-size: 10px;
    letter-spacing: 2px;
}
.error-card {
    margin-top: 20px;
    padding: 40px;
    text-align: center;
    border-radius: 20px;
    background: rgba(80,15,25,0.8);
    border: 1px solid rgba(255,80,80,0.4);
}
.error-icon {
    font-size: 50px;
    color: #ff7373;
}
.error-title {
    margin-top: 10px;
    font-size: 25px;
    font-weight: 800;
    color: #ff8585;
}
.error-text {
    margin-top: 10px;
    color: #d79a9a;
}
@keyframes appear {
    from {
        opacity: 0;
        transform: translateY(15px);
    }
    to {
        opacity: 1;
        transform: translateY(0);
    }
}
.footer {
    padding: 30px;
    text-align: center;
    color: #4e7d88;
    font-size: 10px;
    letter-spacing: 3px;
}
@media(max-width:800px) {
    .title {
        font-size: 32px;
        letter-spacing: 4px;
    }
    .dashboard {
        grid-template-columns: repeat(2,1fr);
    }
    .statistics {
        grid-template-columns: repeat(2,1fr);
    }
}
"""

# ============================================================
# 6. OCEAN BACKGROUND
# ============================================================

ocean_background = """
<div class="bubbles">
    <div class="bubble"></div>
    <div class="bubble"></div>
    <div class="bubble"></div>
    <div class="bubble"></div>
    <div class="bubble"></div>
</div>
"""

# ============================================================
# 7. HEADER
# ============================================================

header = """
<div class="header">
    <div class="title">SONAR</div>
    <div class="subtitle">DEEP SEA INTELLIGENCE COMMAND CENTER</div>
    <div class="radar">
        <div class="radar-line"></div>
        <div class="radar-target"></div>
    </div>
    <div class="status">● SYSTEM ONLINE</div>
</div>
"""

# ============================================================
# 8. DASHBOARD
# ============================================================

dashboard = """
<div class="dashboard">
    <div class="dashboard-card">
        <div class="dashboard-value">ONLINE</div>
        <div class="dashboard-label">SYSTEM STATUS</div>
    </div>
    <div class="dashboard-card">
        <div class="dashboard-value">60</div>
        <div class="dashboard-label">SONAR FEATURES</div>
    </div>
    <div class="dashboard-card">
        <div class="dashboard-value">SVM</div>
        <div class="dashboard-label">AI MODEL</div>
    </div>
    <div class="dashboard-card">
        <div class="dashboard-value">R / M</div>
        <div class="dashboard-label">TARGET CLASSES</div>
    </div>
</div>
"""

# ============================================================
# 9. BUILD GRADIO INTERFACE
# ============================================================

with gr.Blocks(css=css, title="SONAR Deep Sea AI") as app:
    gr.HTML(ocean_background)
    gr.HTML(header)
    gr.HTML(dashboard)

    gr.Markdown("""
    # SONAR SIGNAL ACQUISITION
    Enter the 60 acoustic measurements received from
    the underwater SONAR system.
    """)

    with gr.Group(elem_classes="input-panel"):
        feature_inputs = []

        with gr.Accordion("◈ SIGNAL CHANNEL 01  |  FEATURES 01–10", open=True, elem_classes="signal-group"):
            with gr.Row():
                for i in range(1, 11):
                    component = gr.Number(label=f"F{i:02d}", value=0, precision=5, elem_classes="feature")
                    feature_inputs.append(component)

        with gr.Accordion("◈ SIGNAL CHANNEL 02  |  FEATURES 11–20", open=False, elem_classes="signal-group"):
            with gr.Row():
                for i in range(11, 21):
                    component = gr.Number(label=f"F{i:02d}", value=0, precision=5, elem_classes="feature")
                    feature_inputs.append(component)

        with gr.Accordion("◈ SIGNAL CHANNEL 03  |  FEATURES 21–30", open=False, elem_classes="signal-group"):
            with gr.Row():
                for i in range(21, 31):
                    component = gr.Number(label=f"F{i:02d}", value=0, precision=5, elem_classes="feature")
                    feature_inputs.append(component)

        with gr.Accordion("◈ SIGNAL CHANNEL 04  |  FEATURES 31–40", open=False, elem_classes="signal-group"):
            with gr.Row():
                for i in range(31, 41):
                    component = gr.Number(label=f"F{i:02d}", value=0, precision=5, elem_classes="feature")
                    feature_inputs.append(component)

        with gr.Accordion("◈ SIGNAL CHANNEL 05  |  FEATURES 41–50", open=False, elem_classes="signal-group"):
            with gr.Row():
                for i in range(41, 51):
                    component = gr.Number(label=f"F{i:02d}", value=0, precision=5, elem_classes="feature")
                    feature_inputs.append(component)

        with gr.Accordion("◈ SIGNAL CHANNEL 06  |  FEATURES 51–60", open=False, elem_classes="signal-group"):
            with gr.Row():
                for i in range(51, 61):
                    component = gr.Number(label=f"F{i:02d}", value=0, precision=5, elem_classes="feature")
                    feature_inputs.append(component)

        with gr.Row():
            random_button = gr.Button("⚡ GENERATE TEST SIGNAL", elem_classes="secondary-button")
            clear_button = gr.Button("⌫ CLEAR SIGNAL", elem_classes="secondary-button")

        scan_button = gr.Button("◉  INITIATE SONAR SCAN", elem_classes="scan-button")

    gr.Markdown("# AI DETECTION CENTER")

    result_output = gr.HTML("""
    <div class="result-card">
        <div class="result-icon">◉</div>
        <div class="result-small-title">SONAR CLASSIFICATION</div>
        <div class="result-title">AWAITING SIGNAL</div>
        <div class="result-status">● SYSTEM STANDBY</div>
    </div>
    """)

    gr.HTML("""
    <div class="footer">
        DEEP SEA INTELLIGENCE SYSTEM
        <br><br>
        MACHINE LEARNING • SUPPORT VECTOR MACHINE
        <br><br>
        ROCK / MINE CLASSIFICATION
    </div>
    """)

    scan_button.click(
        fn=predict_sonar,
        inputs=feature_inputs,
        outputs=result_output
    )

    random_button.click(
        fn=generate_random_signal,
        inputs=[],
        outputs=feature_inputs
    )

    clear_button.click(
        fn=clear_signal,
        inputs=[],
        outputs=feature_inputs
    )

# ============================================================
# 10. LAUNCH
# ============================================================

app.launch(share=True, debug=True)
```
